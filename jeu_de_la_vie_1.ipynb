algo du jeu de la vie :

1) creer la grille : 

Fonction creer_grille(nb_lignes, nb_colonnes)

    Déclarer grille comme tableau à deux dimensions

    Pour i allant de 0 à nb_lignes-1 :
        Pour j allant de 0 à nb_colonnes-1 :
            grille[i][j] = 0

    Retourner grille



Cette fonction permet de créer la grille sur laquelle va se dérouler le jeu.
- nb_lignes correspond au nombre de lignes.
- nb_colonnes correspond au nombre de colonnes.
- On parcourt toutes les cases de la grille avec deux boucles.
- On initialise chaque case à 0, donc toutes les cellules sont mortes au départ.
On obtient donc une grille vide.

```python
def creer_grille(nb_lignes, nb_colonnes):

    grille = []

    for i in range(nb_lignes):
        ligne = []

        for j in range(nb_colonnes):
            ligne.append(0)

        grille.append(ligne)

    return grille
```

2) afficher les voisins :


Fonction afficher_voisins(grille, i, j)

    Pour x allant de i-1 à i+1 :
        Pour y allant de j-1 à j+1 :

            Si x >= 0 et x < nb_lignes :
                Si y >= 0 et y < nb_colonnes :

                    Si x != i ou y != j :
                        Afficher grille[x][y]


Cette fonction permet d'afficher les cellules voisines d'une cellule donnée.

On utilise deux boucles pour parcourir les cases autour de la cellule de coordonnées (i,j).

Une cellule possède au maximum 8 voisins :
- 3 au-dessus ;
- 3 en dessous ;
- 1 à gauche ;
- 1 à droite.
On vérifie deux choses :
1. x >= 0 et x < nb_lignes : on vérifie que la ligne existe dans la grille.
2. y >= 0 et y < nb_colonnes : on vérifie que la colonne existe dans la grille.
Enfin, x != i ou y != j permet de ne pas compter la cellule elle-même parmi ses voisins.



```python
def afficher_voisins(grille, i, j):

    nb_lignes = len(grille)
    nb_colonnes = len(grille[0])

    for x in range(i - 1, i + 2):
        for y in range(j - 1, j + 2):

            if x >= 0 and x < nb_lignes:
                if y >= 0 and y < nb_colonnes:

                    if x != i or y != j:
                        print(grille[x][y])
```

3) compter les voisins :

Fonction compter_voisins(grille, i, j)

    nb_voisins = 0

    Pour x allant de -1 à 1 :
        Pour y allant de -1 à 1 :

            Si x != 0 ou y != 0 :

                pos_x = i + x
                pos_y = j + y

                Si pos_x >= 0 et pos_x < nb_lignes :
                    Si pos_y >= 0 et pos_y < nb_colonnes :

                        nb_voisins = nb_voisins + grille[pos_x][pos_y]

    Retourner nb_voisins



Cette fonction permet de compter le nombre de cellules vivantes autour d'une cellule.

On commence par initialiser nb_voisins à zéro.

Ensuite, on parcourt les positions autour de la cellule :
- x varie de -1 à 1 pour les lignes.
- y varie de -1 à 1 pour les colonnes.
On utilise pos_x et pos_y pour obtenir les coordonnées réelles des voisins.

On vérifie ensuite que les coordonnées sont bien dans la grille.

La condition x != 0 ou y != 0 permet d'exclure la cellule centrale.

Enfin, on ajoute la valeur de chaque voisin à nb_voisins.

Pourquoi peut-on directement additionner grille[pos_x][pos_y] ?

Parce que :
- si la cellule est morte, sa valeur est 0 ;
- si la cellule est vivante, sa valeur est 1.
La fonction retourne donc le nombre de voisins vivants.

```python
def compter_voisins(grille, i, j):

    nb_lignes = len(grille)
    nb_colonnes = len(grille[0])

    nb_voisins = 0

    for x in range(-1, 2):
        for y in range(-1, 2):

            if x != 0 or y != 0:

                pos_x = (i + x) % nb_lignes
                pos_y = (j + y) % nb_colonnes

                nb_voisins = nb_voisins + grille[pos_x][pos_y]

    return nb_voisins
```

4) calculer l'etat d'une cellule :

(application des règles du jeu)


Fonction calculer_cellule(etat_cellule, nb_voisins)

    Si etat_cellule = 0 et nb_voisins = 3 :
        Retourner 1

    Sinon si etat_cellule = 1 et
              (nb_voisins = 2 ou nb_voisins = 3) :
        Retourner 1

    Sinon :
        Retourner 0


Cette fonction permet de déterminer si une cellule sera vivante ou morte à la prochaine génération.

Elle prend deux paramètres :
- etat_cellule : l'état actuel de la cellule.
- nb_voisins : le nombre de voisins vivants autour d'elle.
On applique trois règles :
Situation actuelle	Résultat
Cellule morte avec exactement 3 voisins vivants	Elle devient vivante
Cellule vivante avec 2 ou 3 voisins vivants	Elle reste vivante
Toutes les autres situations	Elle devient morte

La fonction retourne donc 1 si la cellule doit être vivante à la prochaine génération, et 0 si elle doit être morte.

```python
def calculer_cellule(etat_cellule, nb_voisins):

    if etat_cellule == 0 and nb_voisins == 3:
        return 1

    elif etat_cellule == 1 and 2 <= nb_voisins <= 3:
        return 1

    else:
        return 0
```

5) calculer la génération suivante : 

Fonction generation_suivante(grille)

    nouvelle_grille = creer_grille(nb_lignes, nb_colonnes)

    Pour i allant de 0 à nb_lignes-1 :
        Pour j allant de 0 à nb_colonnes-1 :

            etat_cellule = grille[i][j]

            nb_voisins = compter_voisins(grille, i, j)

            nouvelle_grille[i][j] =
                calculer_cellule(etat_cellule, nb_voisins)

    Retourner nouvelle_grille




Cette fonction permet de calculer une génération complète à partir de la génération actuelle.

On commence par créer une nouvelle grille remplie de zéros.

Ensuite, on parcourt toutes les cellules de la grille actuelle.

Pour chaque cellule :
1. On récupère son état actuel.
2. On compte ses voisins vivants.
3. On applique les règles du jeu avec calculer_cellule.
4. On enregistre le nouvel état dans la nouvelle grille.
À la fin, on retourne la nouvelle grille.

Pourquoi créer une nouvelle grille ?

C'est un point très important du jeu de la vie.

On ne doit pas modifier directement la grille actuelle pendant le calcul.

Sinon, lorsqu'on calcule les cellules suivantes, certaines utiliseraient déjà les valeurs de la nouvelle génération.

On doit donc conserver la grille actuelle intacte jusqu'à ce que toute la nouvelle génération soit calculée.

```python
def generation_suivante(grille):

    nb_lignes = len(grille)
    nb_colonnes = len(grille[0])

    nouvelle_grille = creer_grille(nb_lignes, nb_colonnes)

    for i in range(nb_lignes):
        for j in range(nb_colonnes):

            etat_cellule = grille[i][j]

            nb_voisins = compter_voisins(grille, i, j)

            nouvelle_grille[i][j] = calculer_cellule(
                etat_cellule, nb_voisins
            )

    return nouvelle_grille
```

C'est le programme principal qui permet de lancer le jeu.
1. On définit les dimensions de la grille.
2. On crée une grille vide.
3. On place les premières cellules vivantes.
4. On affiche la grille initiale.
5. On calcule ensuite les 10 générations suivantes.
6. Après chaque génération, on affiche la nouvelle grille.
La ligne grille = generation_suivante(grille) est particulièrement importante : elle remplace l'ancienne grille par la nouvelle génération une fois que tous les calculs sont terminés.

```python

def creer_grille(nb_lignes, nb_colonnes):
    grille = []

    for i in range(nb_lignes):
        ligne = []

        for j in range(nb_colonnes):
            ligne.append(0)

        grille.append(ligne)

    return grille


def afficher_voisins(grille, i, j):
    nb_lignes = len(grille)
    nb_colonnes = len(grille[0])

    for x in range(i - 1, i + 2):
        for y in range(j - 1, j + 2):

            if x >= 0 and x < nb_lignes:
                if y >= 0 and y < nb_colonnes:

                    if x != i or y != j:
                        print(grille[x][y])


def compter_voisins(grille, i, j):
    nb_lignes = len(grille)
    nb_colonnes = len(grille[0])

    nb_voisins = 0

    for x in range(-1, 2):
        for y in range(-1, 2):

            if x != 0 or y != 0:
                pos_x = (i + x) % nb_lignes
                pos_y = (j + y) % nb_colonnes

                nb_voisins = nb_voisins + grille[pos_x][pos_y]

    return nb_voisins


def calculer_cellule(etat_cellule, nb_voisins):

    if etat_cellule == 0 and nb_voisins == 3:
        return 1

    elif etat_cellule == 1 and 2 <= nb_voisins <= 3:
        return 1

    else:
        return 0


def generation_suivante(grille):
    nb_lignes = len(grille)
    nb_colonnes = len(grille[0])

    nouvelle_grille = creer_grille(nb_lignes, nb_colonnes)

    for i in range(nb_lignes):
        for j in range(nb_colonnes):

            etat_cellule = grille[i][j]

            nb_voisins = compter_voisins(grille, i, j)

            nouvelle_grille[i][j] = calculer_cellule(
                etat_cellule, nb_voisins
            )

    return nouvelle_grille


def initialiser_cellules(grille):
    grille[4][3] = 1
    grille[4][4] = 1
    grille[4][5] = 1

    return grille


def afficher_grille(grille):

    for i in range(len(grille)):
        for j in range(len(grille[0])):

            if grille[i][j] == 1:
                print("O", end=" ")
            else:
                print(".", end=" ")

        print()


def jeu_de_la_vie():

    nb_lignes = 10
    nb_colonnes = 10

    grille = creer_grille(nb_lignes, nb_colonnes)

    grille = initialiser_cellules(grille)

    print("Generation initiale :")
    afficher_grille(grille)

    for generation in range(1, 11):

        grille = generation_suivante(grille)

        print("Generation :", generation)

        afficher_grille(grille)



jeu_de_la_vie()   
```
